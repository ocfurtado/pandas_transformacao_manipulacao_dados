```python
import pandas as pd
import numpy as np
import json
```

```python
with open('dados_locacao_imoveis.json', 'r') as f:
    json_bruto = json.load(f)
```

```python
dados_normalizados = pd.json_normalize(json_bruto, record_path='dados_locacao')
```

```python
dados_normalizados
```

```python
dados_normalizados.info()
```

```python
colunas = list(dados_normalizados.columns)
```

```python
colunas
```

```python
dados_normalizados = dados_normalizados.explode(colunas[1:])
```

```python
dados_normalizados
```

```python
dados_normalizados.reset_index(drop=True, inplace=True)
```

```python
dados_normalizados
```

```python
dados_normalizados['valor_aluguel'] = dados_normalizados['valor_aluguel'].apply(lambda x: x.replace('$ ', '').replace(',', '.').replace('reais', '').strip())
```

```python
dados_normalizados
```

```python
dados_normalizados['valor_aluguel'] = dados_normalizados['valor_aluguel'].astype(np.float64)
```
