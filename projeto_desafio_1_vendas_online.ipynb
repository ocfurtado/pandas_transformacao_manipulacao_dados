```python
import pandas as pd
import numpy as np
import json
```

```python
with open('dados_vendas_clientes.json', 'r') as f:
    json_bruto = json.load(f)
```

```python
dados_normalizados = pd.json_normalize(json_bruto, record_path='dados_vendas')
```

```python
dados_normalizados
```

```python
colunas = list(dados_normalizados.columns)
```

```python
colunas
```

```python
dados_normalizados = dados_normalizados.explode(colunas[1:])
```

```python
dados_normalizados
```

```python
dados_normalizados.reset_index(drop=True, inplace=True)
```

```python
dados_normalizados
```

```python
dados_normalizados.info()
```

```python
dados_normalizados['Valor da compra'] = dados_normalizados['Valor da compra'].apply(lambda x: x.replace('R$ ', '').replace(',', '.').strip())
```

```python
dados_normalizados['Valor da compra'] = dados_normalizados['Valor da compra'].astype(np.float64)
```

```python
dados_normalizados.info()
```

```python
dados_normalizados
```

```python
dados_normalizados['Cliente'] =  (dados_normalizados['Cliente']
 .str.lower()
 .replace(r'[^a-zA-Zá-úÁ-Ú ]', ' ', regex=True))
```
